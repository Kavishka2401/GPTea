### **Conversion of Weather Daily Data to Monthly**

```python
# Ensure time column in dataset is datetime
weather_data_daily['time'] = pd.to_datetime(weather_data_daily['time'])

# Extract Year and Month
weather_data_daily['Year'] = weather_data_daily['time'].dt.year
weather_data_daily['Month'] = weather_data_daily['time'].dt.month  # numeric month
weather_data_daily['Month_name'] = weather_data_daily['Month'].apply(lambda x: pd.to_datetime(str(x), format='%m').strftime('%b'))

# Define aggregation
agg_dict = {
    'temperature_2m_max': 'mean',
    'temperature_2m_min': 'mean',
    'temperature_2m_mean': 'mean',
    'relative_humidity_2m_max': 'mean',
    'relative_humidity_2m_min': 'mean',
    'relative_humidity_2m_mean': 'mean',
    'precipitation_sum': 'sum',
    'rain_sum': 'sum',
    'sunshine_duration': 'sum',
    'wind_speed_10m_max': 'mean',
    'wind_speed_10m_mean': 'mean',
    'et0_fao_evapotranspiration': 'mean'
}

# Group by Year & Month and aggregate
weather_data_monthly = weather_data_daily.groupby(['Year', 'Month', 'Month_name']).agg(agg_dict).reset_index()

# Sort by Year and Month number
weather_data_monthly = weather_data_monthly.sort_values(['Year', 'Month']).reset_index(drop=True)

# Drop numeric Month
weather_data_monthly = weather_data_monthly.drop(columns=['Month'])

# Check result
weather_data_monthly.head()
```

```python
# Save to Google Drive
save_path = '/content/drive/MyDrive/DSGP_DOCS/Datasets/weather_monthly.csv'
weather_data_monthly.to_csv(save_path, index=False)
print("Monthly weather data saved successfully!")
```
